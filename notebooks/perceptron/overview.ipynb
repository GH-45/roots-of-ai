<h1 align="center">Perceptrón Simple</h1>

<div align="center">
    Unidad mínima de decisión en inteligencia artificial, inspirada en el funcionamiento de una neurona biológica. En esencia, es un algoritmo de clasificación binaria capaz de aprender a separar datos en dos categorías trazando una frontera recta entre ellos a partir de ejemplos previos.
</div>

```mermaid
flowchart LR
    X1((X₁))
    X2((X₂))
    Xn((Xₙ))
    Bias((b))

    X1 e1@--- |w₁| Sum(("∑(wᵢ · xᵢ) + b"))
    X2 e2@--- |w₂| Sum
    Xn e3@--- |wₙ| Sum
    Bias e4@--- Sum
    Sum e5@--- fn("f(x)")
    fn e6@--- Out((y))

    classDef input fill:#1E1E2E,stroke:#89B4FA,stroke-width:2px,color:#89B4FA;
    classDef bias fill:#1E1E2E,stroke:#F9E2AF,stroke-width:2px,color:#F9E2AF;
    classDef sum fill:#1E1E2E,stroke:#A6E3A1,stroke-width:2px,color:#A6E3A1;
    classDef activation fill:#1E1E2E,stroke:#CBA6F7,stroke-width:2px,color:#CBA6F7;
    classDef output fill:#1E1E2E,stroke:#F38BA8,stroke-width:2px,color:#F38BA8;

    class X1,X2,Xn input
    class Bias bias
    class Sum sum
    class fn activation
    class Out output

    e1@{ step: 1 }
    e2@{ step: 1 }
    e3@{ step: 1 }
    e4@{ step: 1 }
    e5@{ step: 2 }
    e6@{ step: 3 }
```
